## 6. KPI principal: porcentaje de registros clasificados como `High`

**Nombre:** porcentaje de registros del CSV con `Burnout_Risk_Level = High`. Es la operacionalización del porcentaje de estudiantes clasificados con riesgo alto del documento de referencia, bajo la unidad fila–estudiante declarada.

- **Numerador:** número de filas cuya etiqueta es exactamente `High`.
- **Denominador:** número total de registros del CSV, excluyendo la cabecera. Se mantiene el conjunto completo, sin filtrar otras clases ni eliminar registros para ajustar la línea base.
- **Fórmula:** `KPI = (número de registros High / número total de registros) × 100`.
- **Unidad:** porcentaje (%).
- **Naturaleza:** indicador numérico derivado de una variable categórica; **no es una métrica de desempeño de un modelo**. Las etiquetas se leen del CSV y no son predicciones producidas en esta fase.

**Archivo utilizado:** `dataset/ai_student_impact_dataset (1).csv`, separado por punto y coma (`;`). La sección 6 del documento de `docs` cita `ai_student_impact_dataset.csv`, sin el sufijo `(1)`. En el repositorio revisado se encontró el archivo con sufijo; la diferencia se registra como discrepancia de nombre y no permite afirmar que existan dos datasets distintos.

**Línea base documentada:** 12.487 / 50.000 × 100 = **24,974 %**, aproximadamente **24,97 %**. El código contrasta ambos conteos y el porcentaje exacto, y conserva el resultado real aunque hubiera una diferencia. Usa aritmética decimal para evitar artefactos de coma flotante. La huella SHA-256 permite identificar el contenido del archivo ejecutado.

```python
import csv
import hashlib
import io

csv_path = RAIZ / RUTA_CSV
registros = None
sha256_csv = None
if not csv_path.is_file():
    print('PENDIENTE DE EJECUCIÓN: falta el archivo ' + RUTA_CSV.as_posix())
else:
    contenido_csv = csv_path.read_bytes()
    sha256_csv = hashlib.sha256(contenido_csv).hexdigest()
    lector = csv.DictReader(io.StringIO(contenido_csv.decode('utf-8-sig')), delimiter=';')
    columnas = lector.fieldnames or []
    if 'Burnout_Risk_Level' not in columnas:
        raise ValueError('El CSV identificado no contiene Burnout_Risk_Level; revisar el archivo y el separador ;.')
    registros = list(lector)
    print('CSV utilizado:', RUTA_CSV.as_posix())
    print('Separador: ; | Codificación: UTF-8 (acepta BOM)')
    print('SHA-256:', sha256_csv)
    print('Columnas leídas:', len(columnas))
    print('Registros leídos, sin cabecera:', len(registros))
```

```python
from decimal import Decimal, ROUND_HALF_UP, localcontext
from fractions import Fraction

BASE_TOTAL = 50_000
BASE_HIGH = 12_487
BASE_PORCENTAJE = Decimal('24.974')
seccion_kpi = documento.split('## 8. KPI principal', 1)[1].split('## 9.', 1)[0]
assert all(valor in seccion_kpi for valor in ('12.487', '50.000', '24,974', '24,97')), 'La línea base documentada cambió; revisar las referencias.'
assert Fraction(BASE_HIGH * 100, BASE_TOTAL) == Fraction(BASE_PORCENTAJE)

def entero_es(valor):
    return f'{valor:,}'.replace(',', '.')

def decimal_es(valor):
    return format(valor, 'f').replace('.', ',')

resultado_kpi = None
if registros is None:
    print('KPI pendiente: CSV no disponible. La línea base de docs es una referencia, no una salida calculada.')
else:
    total = len(registros)
    n_high = sum(fila['Burnout_Risk_Level'] == 'High' for fila in registros)
    etiquetas = {fila['Burnout_Risk_Level'] for fila in registros}
    clases_esperadas = {'Low', 'Medium', 'High'}
    n_etiquetas_invalidas = sum(fila['Burnout_Risk_Level'] not in clases_esperadas for fila in registros)
    print('Etiquetas presentes (sin modificar):', sorted(etiquetas, key=lambda valor: str(valor)))
    print('Etiquetas vacías o fuera de Low/Medium/High:', n_etiquetas_invalidas)
    if total == 0:
        print('KPI no definido: el CSV no contiene registros; el denominador es cero.')
    else:
        fraccion_exacta = Fraction(n_high * 100, total)
        with localcontext() as contexto:
            contexto.prec = 28
            porcentaje = Decimal(n_high) * Decimal(100) / Decimal(total)
            porcentaje_dos_decimales = porcentaje.quantize(Decimal('0.01'), rounding=ROUND_HALF_UP)
        coincide_base = (total == BASE_TOTAL and n_high == BASE_HIGH
                         and fraccion_exacta == Fraction(BASE_PORCENTAJE))
        resultado_kpi = {'total': total, 'high': n_high, 'porcentaje': porcentaje,
                         'redondeado': porcentaje_dos_decimales, 'coincide_base': coincide_base,
                         'etiquetas_invalidas': n_etiquetas_invalidas}
        print('Total de registros (denominador):', entero_es(total))
        print('Registros High (numerador):', entero_es(n_high))
        print('Porcentaje sin redondear a dos decimales:', decimal_es(porcentaje) + ' %')
        print('Fracción porcentual exacta:', str(fraccion_exacta) + ' %')
        print('Porcentaje redondeado a dos decimales:', decimal_es(porcentaje_dos_decimales) + ' %')
        print('Referencia de docs: 12.487 / 50.000 × 100 = 24,974 % ≈ 24,97 %')
        if coincide_base:
            print('Contraste: COINCIDE con los conteos y el porcentaje de la línea base documentada.')
        else:
            print('DISCREPANCIA con docs: se conserva el cálculo del CSV identificado, sin ajustar registros ni etiquetas.')
            print('Diferencia de total:', total - BASE_TOTAL, '| Diferencia de High:', n_high - BASE_HIGH)
            print('Diferencia porcentual (puntos porcentuales):', decimal_es(porcentaje - BASE_PORCENTAJE))
        if n_etiquetas_invalidas:
            print('Revisar etiquetas en Fase 2: el denominador incluye todos los registros, también los de etiqueta inválida.')
```

## 7. Interpretación del KPI

El **24,97 % de la línea base documentada**, cuando es reproducido por las celdas anteriores, describe la proporción de registros etiquetados `High` en el CSV: aproximadamente uno de cada cuatro. No es una prevalencia clínica de burnout ni una estimación representativa de todos los universitarios. Si se detectara una discrepancia, prevalece para esta ejecución el valor calculado y se debe revisar la documentación antes de adoptar su línea base.

Es una fotografía descriptiva del conjunto, **no una reducción del riesgo**. No hay una comparación antes–después de una intervención. Tampoco informa cuántas predicciones acertaría un modelo: el numerador cuenta etiquetas existentes.

Reducir la proporción `High` respecto de esta línea base es una **meta futura**. Evaluarla requeriría una intervención real, mediciones comparables, una definición estable de la etiqueta y del denominador, períodos definidos y una población comparable. Además, atribuir una eventual variación a la intervención exigiría un diseño de evaluación apropiado. Una menor cantidad de predicciones `High` por cambiar un clasificador no demostraría por sí sola una mejora real.
